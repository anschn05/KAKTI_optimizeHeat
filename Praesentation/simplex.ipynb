$max \,c^Tx$

$ Ax \leq b $

## Recipe:
1. choose column, that improves max-Function
2. look, which row sets border the lowest
3. choose pivot-element
4. norm pivot-row
5. get other entries in this column to 0
6. repeat :)

```julia
using LinearAlgebra

function simplex(A, b, c)

    m, n = size(A)
    # m Anzahl Nebenbedingungen
    # n Anzahl Entscheindungsvariablen

    # Tableau:
    # [ A  I  b ]
    # [-c  0  0 ]

    tableau = [
        A Matrix(I, m, m) b;        # Matrix I für schlupfvariablen
        -c' zeros(1, m) 0
    ]


# x1  x2  s1  s2  s3 | RHS
# 1   1   1   0   0  | 4
# 1   0   0   1   0  | 2
# 0   1   0   0   1  | 3
# -------------------------
# -3 -2   0   0   0  | 0

    while true

        # 1. Entering variable:
        # most negative value in objective row
        objective_row = tableau[end, 1:end-1]

        if all(objective_row .>= 0)
            break   # optimal solution found
        end

        pivot_col = argmin(objective_row)       # chooses -3 -> x1

        # 2. Leaving variable:
        # ratio test
        ratios = fill(Inf, m)

        for i in 1:m
            if tableau[i, pivot_col] > 0
                ratios[i] =
                    tableau[i, end] / tableau[i, pivot_col]
            end
        end

        pivot_row = argmin(ratios)              # ratios = [4,2,inf] -> pivot-row(2) = 2. zeile

        if ratios[pivot_row] == Inf
            error("Problem is unbounded")
        end

        # 3. Pivot - wollen pivot spalte auf 0 bringen, bis auf pivot-eintrag in zeile 2
        pivot = tableau[pivot_row, pivot_col]

        tableau[pivot_row, :] ./= pivot

        for i in 1:m+1
            if i != pivot_row
                tableau[i, :] .-=
                    tableau[i, pivot_col] .* tableau[pivot_row, :]
            end
        end
    end

    return tableau
end

```

```julia
A = [
    1.0 1.0
    1.0 0.0
    0.0 1.0
]

b = [4.0, 2.0, 3.0]

c = [3.0, 2.0]

T = simplex(A, b, c)

display(T)
```
